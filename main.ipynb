```python
import pandas as pd

# 1. قراءة الملف الأصلي أو عينة الملف لديك
input_file = 'archive/Automated_Traffic_Volume_Counts.csv' # أو اسم ملف العينة لو شغال عليه حالياً

# 2. تحديد أفضل الـ Features التي اتفقنا عليها فقط
best_features = [
    'Boro', 'Yr', 'M', 'D', 'HH', 'MM', 
    'Vol', 'SegmentID', 'WktGeom', 'Direction'
]

# قراءة الملف بالأعمدة المختارة فقط
df_cleaned = pd.read_csv(input_file, usecols=lambda col: col in best_features)

# 3. تنظيف البيانات السريع (حذف الصفوف التي بها قيم ناقصة في الأساسيات)
df_cleaned = df_cleaned.dropna(subset=['Boro', 'Yr', 'M', 'D', 'HH', 'Vol'])
df_cleaned = df_cleaned[df_cleaned['Vol'] >= 0]
df_cleaned = df_cleaned.drop_duplicates()

# 4. حفظ الملف النهائي النظيف
output_file = 'final_cleaned_traffic_data.csv'
df_cleaned.to_csv(output_file, index=False)

print(f"✅ تم تصفية وحفظ الملف بنجاح! إجمالي الأعمدة المختارة: {df_cleaned.shape[1]}")
print(df_cleaned.head())
```
